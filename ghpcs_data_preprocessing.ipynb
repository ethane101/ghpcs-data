# Loading in the Data Sets

```python
with open('/drive/My Drive/GHP CS/research-project/raw-data/compressed-data.pkl', 'rb') as compdata:
  cnv_df = pickle.load(compdata)
  methy_df = pickle.load(compdata)
```

```python
sample_dict = pd.read_csv('https://raw.githubusercontent.com/ethane101/ghpcs-data/refs/heads/main/sample_dict.csv', index_col = 0).to_dict()
```

```python
surv_df = pd.read_csv('https://raw.githubusercontent.com/ethane101/ghpcs-data/59246d4717a665523fbd5dc3e0790d4d3232ea1b/raw-data/54814661_survival_ACC.csv')
surv_df.drop(columns='sample_name', inplace=True)

# sample_name is the patient, event_observed represents whether the patient died (1 = died, 0 = alive), survival_times represents how long they were observed/treated for
# 177 patient samples (separate patients)
```

```python
#cnv_df = pd.read_csv('https://raw.githubusercontent.com/ethane101/ghpcs-data/59246d4717a665523fbd5dc3e0790d4d3232ea1b/raw-data/ACC_CNV_aligned.csv')
cnv_df.rename(columns={'Unnamed: 0':''}, inplace=True)
cnv_df = cnv_df.transpose()
cnv_df.columns = cnv_df.iloc[0]
cnv_df.drop('', axis='index', inplace=True)
cnv_df.reset_index(inplace=True, drop=True)
cnv_df.head()
```

```python
#methy_df = pd.read_csv('https://raw.githubusercontent.com/ethane101/ghpcs-data/refs/heads/main/ACC_Methy_aligned.csv')
methy_df.rename(columns={'Unnamed: 0':''}, inplace=True)
methy_df.set_index('', inplace=True)
methy_df = methy_df.transpose()
methy_df.reset_index(inplace=True, drop=True)
```

```python
mrna_df = pd.read_csv('https://raw.githubusercontent.com/ethane101/ghpcs-data/1a4a7ad904acc5923ac8d7e22997e66d6a0fec74/ACC_mRNA_aligned.csv')
mrna_df.rename(columns={'Unnamed: 0':''}, inplace=True)
mrna_df.set_index('', inplace=True)
mrna_df = mrna_df.transpose()
mrna_df.reset_index(inplace=True, drop=True)
```

```python
mirna_df = pd.read_csv('https://raw.githubusercontent.com/ethane101/ghpcs-data/1a4a7ad904acc5923ac8d7e22997e66d6a0fec74/ACC_miRNA_aligned.csv')
mirna_df.rename(columns={'Unnamed: 0':''}, inplace=True)
mirna_df = mirna_df.transpose()
mirna_df.columns = mirna_df.iloc[0]
mirna_df.drop('', axis='index', inplace=True)
mirna_df.reset_index(inplace=True, drop=True)
```

```python
dfs = [surv_df, cnv_df, methy_df, mrna_df, mirna_df]
dfs[0].name = 'surv_df'
dfs[1].name = 'cnv_df'
dfs[2].name = 'methy_df'
dfs[3].name = 'mrna_df'
dfs[4].name = 'mirna_df'
```

# Normalizing the Data

```python
def normalize_df(df):
  for column in df.columns:
    df[column] = (df[column] - df[column].min()) / (df[column].max() - df[column].min())
```

```python
for df in dfs:
  normalize_df(df)
```

# Exporting the Data

```python
with open('processed-data.pkl', 'wb') as procdata:
  for df in dfs:
    pickle.dump(df, procdata)
```
